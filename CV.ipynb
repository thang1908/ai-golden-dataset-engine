```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from skimage import data
```

```python
img_rgb = data.astronaut()

print(img_rgb.shape)
```

```python
plt.figure(figsize=(6,6))

plt.imshow(img_rgb)

plt.axis("off")

plt.show()
```

```python
img = cv2.cvtColor(
    img_rgb,
    cv2.COLOR_RGB2BGR
)
```

```python
pixel = img[100,100]

print(pixel)
```

```python
gray = cv2.cvtColor(
    img,
    cv2.COLOR_BGR2GRAY
)
```

```python
print(gray.shape)
```

```python
plt.imshow(
    gray,
    cmap="gray"
)

plt.axis("off")
plt.show()
```

```python
small = cv2.resize(
    img,
    (256,256)
)
```

```python
print(small.shape)
```

```python
h, w = img.shape[:2]

center = (
    w//2,
    h//2
)


M = cv2.getRotationMatrix2D(
    center,
    45,
    1
)


print(M)
```

```python
rotated = cv2.warpAffine(
    img,
    M,
    (w,h)
)
```

```python
plt.figure(figsize=(10,5))


plt.subplot(1,2,1)

plt.imshow(
    cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
)

plt.title("Original")

plt.axis("off")



plt.subplot(1,2,2)

plt.imshow(
    cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)
)

plt.title("Rotated 45 degree")

plt.axis("off")


plt.show()
```

```python
plt.imshow(
    gray,
    cmap="gray"
)

plt.axis("off")

plt.show()
```

```python
threshold_value = 120
ret, binary = cv2.threshold(
    gray,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)
```

```python
plt.figure(figsize=(10,4))


plt.subplot(1,2,1)

plt.imshow(
    gray,
    cmap="gray"
)

plt.title("Gray image")

plt.axis("off")



plt.subplot(1,2,2)

plt.imshow(
    binary,
    cmap="gray"
)

plt.title("Threshold")

plt.axis("off")


plt.show()
```

```python
identity = np.array(
[
[0,0,0],
[0,1,0],
[0,0,0]
]
)
```

```python
result_identity = cv2.filter2D(
    img,
    -1,
    identity
)
```

```python
plt.imshow(
    cv2.cvtColor(
        result_identity,
        cv2.COLOR_BGR2RGB
    )
)

plt.show()
```
